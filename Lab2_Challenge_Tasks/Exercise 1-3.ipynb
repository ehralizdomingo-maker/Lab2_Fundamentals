```python
import random

surname = "Domingo"
SEED_NUM = 0

random.seed(SEED_NUM)

sensor_data = [random.randint(0, 100) for _ in range(5)]

print("Generated Sensor Data:", sensor_data)

valid_results = []
invalid_results = []

for reading in sensor_data:
    if isinstance(reading, int):
        valid_results.append(reading)
    else:
        invalid_results.append(reading)

print("Valid Results:", valid_results)
print("Invalid Results:", invalid_results)
```

```python
surname = "Domingo"
SEED_NUM = 0

signal = surname + str(SEED_NUM)

print("Generated Signal:", signal)

processed_signal = signal.strip().upper()

print("Processed Signal:", processed_signal)

for char in processed_signal:
    if char.isalpha():
        print(char, "- Letter")
    elif char.isdigit():
        print(char, "- Number")
    elif char.isspace():
        print(char, "- Space")
    else:
        print(char, "- Other")
```

```python
surname = "Domingo"
SEED_NUM = 0

password = surname.upper() + str(SEED_NUM)
attempt_limit = SEED_NUM + 3

print("Generated Password:", password)
print("Attempt Limit:", attempt_limit)

attempts = 0
access_granted = False

while attempts < attempt_limit:
    user_input = input("Enter Password: ")
    attempts += 1

    if user_input == password:
        access_granted = True
        print("Access Granted")
        break
    else:
        print("Access Denied")

if access_granted:
    print("Final System State: UNLOCKED")
else:
    print("Final System State: LOCKED")
```
